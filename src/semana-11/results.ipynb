```python
import pandas as pd

# Lê os três CSVs
simple_df = pd.read_csv("C:/projects/IC/LIPAI - Onboarding/src/semana-11/thresholding_methods/simple_thresholding/dice_results.csv")
adaptive_df = pd.read_csv("C:/projects/IC/LIPAI - Onboarding/src/semana-11/thresholding_methods/adaptive_thresholding/dice_results.csv")
otsu_df = pd.read_csv("C:/projects/IC/LIPAI - Onboarding/src/semana-11/thresholding_methods/otsu_riddler/dice_results.csv")

# Faz merge (junção) pelos nomes das imagens
merged_df = simple_df.merge(adaptive_df, on="imagem")
merged_df = merged_df.merge(otsu_df, on="imagem")

# Salva em um novo CSV
merged_df.to_csv("dice_results_all_methods.csv", index=False)
```
